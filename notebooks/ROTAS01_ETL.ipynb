```python
import pandas as pd

# Supondo que o dataset já esteja carregado em df

# Colunas categóricas que precisam de padronização
colunas_para_padronizar = ['category', 'sub_category', 'region', 'segment', 'ship_mode', 'market', 'market2', 'state']

# 1️⃣ Verificar valores únicos antes da padronização
for coluna in colunas_para_padronizar:
    print(f"Valores únicos em {coluna} antes da padronização:")
    print(df[coluna].unique())
    print("-"*50)

# 2️⃣ Padronizar para letras minúsculas
for coluna in colunas_para_padronizar:
    df[coluna] = df[coluna].str.lower()

# 3️⃣ Conferir valores únicos após padronização
for coluna in colunas_para_padronizar:
    print(f"Valores únicos em {coluna} após padronização:")
    print(df[coluna].unique())
    print("-"*50)
```

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
# URL da página
url = "https://en.wikipedia.org/wiki/List_of_supermarket_chains"

# Cabeçalho para evitar erro 403
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/116.0.5845.110 Safari/537.36"
}

# Baixar a página
resposta = requests.get(url, headers=headers)

# Transformar o conteúdo HTML em objeto BeautifulSoup
soup = BeautifulSoup(resposta.text, 'html.parser')

# Encontrar todas as tabelas da página
tabelas = soup.find_all('table')

# Mostrar quantas tabelas foram encontradas
print(len(tabelas))
```

```python
# Vamos olhar as primeiras linhas de cada tabela para identificar
for i, tabela in enumerate(tabelas):
    print(f"Tabela {i}")
    linhas = tabela.find_all('tr')
    for linha in linhas[:3]:  # mostrar apenas as 3 primeiras linhas de cada tabela
        colunas = linha.find_all(['th', 'td'])
        print([c.text.strip() for c in colunas])
    print("-"*50)
```

```python
# Selecionar a tabela 1
tabela_multinacional = tabelas[1]

# Extrair todas as linhas
linhas = tabela_multinacional.find_all('tr')

# Criar lista de listas para os dados
dados = []
for linha in linhas:
    colunas = linha.find_all(['th', 'td'])
    dados.append([c.text.strip() for c in colunas])

# Transformar em DataFrame
import pandas as pd

# A primeira linha é o cabeçalho
df_multinacional = pd.DataFrame(dados[1:], columns=dados[0])

# Mostrar as primeiras linhas
df_multinacional.head()
```

```python
import pandas as pd

# Copiar tabela fato
fato_multinacional = df_multinacional[['Company', 'Headquarters', 'Number of locations', 'Number of employees']].copy()

# Criar ID
fato_multinacional['id_fato'] = range(1, len(fato_multinacional)+1)

# Limpar 'Number of employees'
fato_multinacional['Number of employees'] = (
    fato_multinacional['Number of employees']
    .str.replace('+','', regex=False)
    .str.replace(',','', regex=False)
    .str.extract(r'(\d+)')  # usar raw string r''
)

# Limpar 'Number of locations'
fato_multinacional['Number of locations'] = (
    fato_multinacional['Number of locations']
    .str.replace(',','', regex=False)
    .str.extract(r'(\d+)')
)

# Converter para int, preenchendo NaN com 0 ou outro valor apropriado
fato_multinacional['Number of employees'] = fato_multinacional['Number of employees'].fillna(0).astype(int)
fato_multinacional['Number of locations'] = fato_multinacional['Number of locations'].fillna(0).astype(int)

# Reordenar colunas
fato_multinacional = fato_multinacional[['id_fato', 'Company', 'Headquarters', 'Number of locations', 'Number of employees']]

fato_multinacional.head()
```
